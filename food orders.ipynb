### Data-Quality Assessment Summary

The assessment identified the following data-quality issues in the food orders dataset:

**1. Missing values**

Several columns contain missing values. The `Discounts and Offers` column has 242 missing values (23.96%), while `Payment Processing Fee` has 51 missing values and `Payment Method` has 50 missing values.

**Why it matters:** Missing financial information can affect revenue and cost calculations. Missing payment methods can also reduce the accuracy of payment-preference analysis. Missing discount information makes it difficult to measure promotional performance.

**2. Duplicate records**

The dataset contains 10 fully duplicated rows. It has 1,010 rows but only 1,000 unique `Order ID` values. The repeated order IDs correspond to identical records.

**Why it matters:** Duplicate records can inflate order counts and cause incorrect calculations of total order value, commissions, fees, and refunds. These records should be verified against the source before duplicates are removed.

**3. Unexpected timestamp data types**

The `Order Date and Time` and `Delivery Date and Time` columns are stored as text (`object`) rather than datetime values. However, all timestamps were successfully parsed into datetime values.

**Why it matters:** Datetime conversion is necessary to calculate delivery duration accurately and analyze order trends by date, time, or month.

**4. Missing categorical information**

The `Payment Method` column has 50 missing values, and `Discounts and Offers` has 242 missing values.

**Why it matters:** Missing categories can lead to incomplete reports about customer payment preferences and discount usage. Missing values should be investigated rather than automatically assumed to represent a particular payment method or discount.

**5. Financial and delivery-duration validation**

No negative values were found in the financial columns, and no records had a delivery timestamp earlier than the order timestamp. Delivery durations ranged from 30 to 119 minutes, with an average of approximately 73.61 minutes.

**Why it matters:** Negative financial amounts or delivery durations could indicate data-entry or processing errors. Although none were found in these checks, financial values should still be compared with business rules to identify other possible inconsistencies.

### Conclusion

The dataset is usable for further analysis after appropriate data-quality checks and preparation. The main concerns are missing values, duplicate records, and timestamp data types. ### Data-Cleaning Rules

| Data-quality issue | Your chosen handling rule | Reason / business assumption |
|---|---|---|
| Exact duplicate rows | Remove fully duplicated rows after inspecting them. Keep one copy of each record. | Exact duplicates can inflate order counts, revenue, commissions, and other financial totals. |
| Repeated Order ID values | Check repeated IDs after removing exact duplicates. Investigate any remaining repeated IDs instead of automatically deleting them. | Multiple records with the same ID may represent different transactions or processing stages. |
| Missing Order Value | Retain missing values as null. Exclude them from order-value calculations and report how many values are missing. | A missing order value does not mean the order was worth zero. Filling it with zero could underestimate sales. |
| Missing Delivery Fee | Retain missing values as null. | A missing delivery fee does not necessarily mean delivery was free. |
| Missing Commission Fee | Retain missing values as null. | Assuming zero commission could underestimate platform earnings. |
| Missing Payment Processing Fee | Retain missing values as null. | Replacing missing fees with zero could underestimate payment-processing costs. |
| Missing Refunds/Chargebacks | Retain missing values as null. | A missing refund amount does not necessarily mean that no refund or chargeback occurred. |
| Missing Payment Method | Replace missing values with `Unknown`. | This makes missing payment information visible without guessing which method was used. |
| Missing Discounts and Offers | Replace missing values with `Not Specified`. | Missing offer details do not prove that no discount was applied. |
| Invalid timestamps | Convert timestamp columns using `pd.to_datetime()` with `errors="coerce"`. Unparseable values become `NaT` and are flagged for investigation. | Invalid timestamps cannot reliably support time-based analysis. |
| Impossible delivery durations | Calculate duration only when both timestamps are valid. If delivery occurs before the order, set the calculated duration to missing and flag the record for investigation. | Negative delivery durations are not logically valid. Retaining the order preserves potentially useful financial information. |
| Financial columns with invalid numeric values | Convert financial columns using `pd.to_numeric()` with `errors="coerce"` and investigate values that become missing. | Numeric values are needed for calculations. Invalid text should not be treated as a valid financial amount. |
